```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import geopandas as gpd
from shapely.geometry import Polygon, Point
import numpy as np
import matplotlib.pyplot as plt
import time
from matplotlib_scalebar.scalebar import ScaleBar
import matplotlib.patches as mpatches
import matplotlib.cm
from matplotlib.patches import Rectangle
```

```python
#just using this to grab an up to date list of sites
temp = pd.read_csv('C:/Users/sarah/Documents/GitHub/ORKA_StatusReport_FollowupPaper/StatusReport_FollowupPaper_Data/MasterSites_WithTempMetrics_20132024.csv')
site_list = temp.Site.values.tolist()
```

```python
#For Reef Scale Measurements
data = []
for site in site_list:
    print(site)
    #Lookout not used since it was surveyed only in 2022, not 2023
    if site == "Cape Lookout":
        data.append([site, "NA","NA"])
    else:
        site_str = site.replace(" ","")
        pathname = "C:/Users/sarah/Documents/GitHub/ORKA_StatusReport_FollowupPaper/StatusReport_FollowupPaper_Data/KelpWatchData_Dec2025/KW_" + site_str + "_ReefScale.csv"
        d = pd.read_csv(pathname)
        dearly = d.loc[(d.year.between(2010,2013)) & (d.quarter == "max")].kelp_area_m2.median()
        dearly2 = d.loc[(d.year.between(2013,2016)) & (d.quarter == "max")].kelp_area_m2.median()
        dlate = d.loc[(d.year.between(2020,2023)) & (d.quarter == "max")].kelp_area_m2.median()
        d2010 = d.loc[(d.year==2010) & (d.quarter == "max")].kelp_area_m2.median()
        d2022 = d.loc[(d.year==2022) & (d.quarter == "max")].kelp_area_m2.median()
        d2024 = d.loc[(d.year==2024) & (d.quarter == "max")].kelp_area_m2.median()
        data.append([site, dearly,dearly2, dlate, d2010, d2022, d2024])
```

```python
#Note Sara. I added the 0.01 to the Med 2023 column so you wouldn't get zero in the Pacific City Median for 2023
df = pd.DataFrame(data,columns = ["Site","Med1013","Med1316","Med2023","Max2010","Max2022","Max2024"])
df2 = df.iloc[1:,]
df2 = df2.replace({"Med1316": 0}, 1)
df2["Percentage_Of1013"] = (((df2["Med2023"])/df2["Med1013"])*100).astype(float)
df2["Percentage_Of1316"] = (((df2["Med2023"])/df2["Med1316"])*100).astype(float)
df2["Log_Percentage1013"] = np.log(df2.Percentage_Of1013)
df2["Log_Percentage1316"] = np.log(df2.Percentage_Of1316)
df2
```
